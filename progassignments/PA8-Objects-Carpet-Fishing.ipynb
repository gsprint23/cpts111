### Overview and Requirements
Write a program (carpet_fishing.py) to implement [Dilbert's Carpet Fishing Game](http://dilbert.com/strip/2007-09-30).

<img src="https://raw.githubusercontent.com/gsprint23/cpts111/master/progassignments/figures/dilbert_carpet_fishing.jpg" width="700"/>
(image from [http://dilbert.com/strip/2007-09-30](http://dilbert.com/strip/2007-09-30))

Let's call Dilbert's cubicle floor "Carpet Sea". Carpet Sea is divided into a grid of `N` x `N` cells. Your program should randomly select a cell in the carpet grid to place a fish and then prompt the user to enter the cell in which they want to drop their fishing line. Every `M` minutes, reel in the fishing line and check if the user caught a fish (i.e. they dropped their line in the same cell as the randomly placed fish). Repeat this process:
1. Randomly place a fish
1. Prompt for a fishing line location
1. Wait `M` minutes
1. Reel in the line and check if the fish was caught

until the user wants to quit.

#### Program Details
Your code should include `N` and `M` variables and be able to handle different integer values of `N` (2 <= `N` <= 9) and `M` (`M` >= 15); however, for simplicity, you may hard-code `N = 2` and `M = 60` in your program instead of prompting the user for values for `N` and `M`. If you do this, Carpet Sea will be a 2 x 2 grid and the user will reel in the line once every hour.

##### Simulation of Time
To save on time, we are going to simulate `M` minutes by `S` real-time seconds, where 1 real-time second is equivalent to 1/60 simulated minutes:

$$S = M\frac{1}{60}$$

For example, in the original comic, Dilbert checks his fishing line once every hour (`M` = 60 minutes). So our program will check the fishing line once every second (`S` = 1) in real-time: 

$$S = 60 \frac{1}{60} = 1$$ 

We can have our program "wait" for `S` seconds by importing the `time` module and calling `time.sleep(S)`.

##### Carpet Sea
The underlying representation of Carpet Sea should include the following classes:
1. `Coordinate`: represents the coordinates of a cell in the carpet grid. `Coordinate` should contain the following attributes and methods (at a minimum):
    1. `row`: a row location (an integer)
    1. `col`: a column location (an integer)
    1. `__init__()`: accepts values to assign to the attributes.
    1. `__str__()`: returns a string representation of a `Coordinate`, i.e. (0, 0)
1. `Cell`: represents a cell in the carpet grid. `Cell` should contain the following attributes and methods (at a minimum):
    1. `location_coords`: the location of this cell on the board (a `Coordinate` object).
    1. `contains_line`: whether or not this cell contains the user's fishing line (a `bool`)
    1. `fish`: the fish contained in this cell (a string)
    1. `__init__()`: accepts values to assign to the attributes.
    1. `__str__()`: returns a string representation of a `Cell`, i.e. return the contents of this cell (e.g. " " (nothing), "F" (fish), "`*`" (line), "F`*`" (fish and line)
1. `CarpetSea`: represents the carpet grid. `CarpetSea` should contain the following attributes and methods (at a minimum):
    1. `N`: dimension of board (an integer)
    1. `grid`: the `N` x `N` grid (a 2-Dimensional list of `Cell` objects)
    1. `available_fish`: species of fish in the Carpet Sea (a list of strings)
    1. `__init__()`: accepts a value to assign to the attribute `N`.
    1. `__str__()`: returns a string representation of a `CarpetSea`, i.e. display the organized contents of each cell. Rows and columns should be labeled with indices.
    1. `randomly_place_fish()`: randomly selects coordinates of a cell and randomly selects a fish from the `available_fish` list attribute. Marks the cell as containing the fish.
    1. `drop_fishing_line()`: accepts the location of the user's fishing line (a `Coordinate` object). Marks the cell as containing the line.
    1. `check_fish_caught()`: If the cell containing the fishing line also contains a fish, returns the fish. Otherwise, return `False`.
 
##### Carpet Fishing Game Stats
Keep track of the game progress in a `GameStats` class. When the user wants to quit playing Carpet Fishing, report the user's game stats, including:
* Total number of fish caught
* Total (simulated) time played in terms of hours

Be sure to define `__init__()` and `__str__()` methods for `GameStats`!

Display instructions at the beginning of the program, including a description of the fish in Carpet Sea.

Prompt the user for the following values during game play:

1. Menu choice: after checking if a fish was caught, prompt the user to determine if they want to keep fishing. **Validate the user enters "y" to keep playing or "q" to quit.**

Display the Carpet Sea grid after checking if a fish was caught.
